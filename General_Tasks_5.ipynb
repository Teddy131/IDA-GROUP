```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
#reading the csv-file
data = pd.read_csv()

```

```python
#defining feature and target vektor

X = data[['days', 'fuel', 'engine']]

y = data['Fehlerhaft_Fahrleistung']
```

```python
#Train-Test-Split for evaualtion

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
#perpairing the data for training

# the engine colum has 3 vlaues: small, medium and large. the computer cannot work with those, so we need to change them to numbers via onehot encoding

encoder = OneHotEncoder(sparse_output=False, drop='first')

encoder.fit(x_train[['engine']])
#the train and the testvektor have now encoded values


encoded_engine_train = encoder.transform(x_train[['engine']])
encoded_engine_test = encoder.transform(x_test[['engine']])
# the new vektors must be combined the other features

X_train = np.column_stack([x_train['days'].to_numpy(), 
                   x_train['fuel'].to_numpy(), encoded_engine_train])

X_test = np.column_stack([x_test['days'].to_numpy(), 
                   x_test['fuel'].to_numpy(), encoded_engine_test])
```

```python
#Model
model = LinearRegression()
model.fit(X=X_train, y=y_train)
```
