# Task 6: hit-and-run accident
On 11.08.2010, there was a hit-and-run accident. The license plate of the car involved is unknown. The police have asked for your assistance, as you work for the Federal Motor Transport Authority, to find out where the vehicle with **body part number** “K5-112-1122-79” was registered.

**Strategy**

1- Check the file names manually and see the logic.

2- Check the files Bestandteile_Fahrzeuge_X, find the vehicle ID through the given body part ID.

3- Cross check in the matching Fahrzeuge_X file, might be useful later too.

4- Check the registration data

5- Check the location data

**Imports:**

```python
import pandas as pd
```

**Check, if the file contains the information we need**

```python
df = pd.read_csv("data/Fahrzeug/Bestandteile_Fahrzeuge_OEM1_Typ11.csv", sep = ";")
df.head()
```

**-> We understand that we need to check for ID_Karosserie**

```python
body_part_number = "K5-112-1122-79"

bestandteile_files = ["OEM1_Typ11.csv", "OEM1_Typ12.csv", "OEM2_Typ21.csv", "OEM2_Typ22.csv"]

for i in bestandteile_files:
    body_parts = pd.read_csv("data/Fahrzeug/Bestandteile_Fahrzeuge_" + i, sep = ";")
    check = body_parts[body_parts["ID_Karosserie"] == body_part_number]
    if len(check) > 0:
        print(check)
        break
```

**-> We have found the matching row**

```python
id_vehicle = check["ID_Fahrzeug"].values[0]
print(id_vehicle)
```

```python
vehicles = pd.read_csv("data/Fahrzeug/Fahrzeuge_OEM1_Typ12.csv", sep = ";")
vehicles.head()
```

```python
check_vehicle = vehicles[vehicles["ID_Fahrzeug"] == id_vehicle]
check_vehicle
```

**-> The vehicle was produced before 2010, so the information we found is probably correct.**

```python
registration = pd.read_csv("data/Zulassungen/Zulassungen_alle_Fahrzeuge.csv", sep = ";")
registration.head()
```

```python
find_registration = registration[registration["IDNummer"] == id_vehicle] 
find_registration
```

```python
vehicle_registered_in = find_registration["Gemeinden"].values[0]
vehicle_registered_in
```

**-> Location**

```python
geodata = pd.read_csv("data/Geodaten/Geodaten_Gemeinden_v1.2_2017-08-22_TrR.csv", sep = ";")
geodata
```

```python
aschersleben_details = geodata[geodata["Gemeinde"] == "ASCHERSLEBEN"]
aschersleben_details
```

```python
aschersleben_postcode = aschersleben_details["Postleitzahl"].values[0]
aschersleben_longitude = aschersleben_details["Laengengrad"].values[0]
aschersleben_latitude = aschersleben_details["Breitengrad"].values[0]
```

```python
print(f"The vehicle {id_vehicle} containing body part with id {body_part_number} is registered in {vehicle_registered_in}"
      f", post code: {aschersleben_postcode}, longitude: {aschersleben_longitude} latitude: {aschersleben_latitude}")
```
